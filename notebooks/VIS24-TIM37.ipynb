**Zaključak:** Ovaj grafikon prikazuje odnos između broja streamova pjesama i njihovog prisustva na
Spotify playlistama (označeno kao "in_spotify_pl"). Očigledno je da kako se broj
prisustava pjesme na playlistama povećava, tako se generalno povećava i broj
streamova, ali ovaj odnos nije strogo linearan.
Većina tačaka grupisana je na donjem lijevom dijelu grafikona, gdje pjesme s manjim

koje su izolirane i visoko iznad drugih na grafikonu), postoji i mnogo pjesama s visokim

Zaključno, dok postoji pozitivna korelacija između prisustva pjesama na Spotify
playlistama i broja streamova, ovaj odnos nije uvijek dosljedan, te postoji varijabilnost
koja može biti pod utjecajem drugih faktora kao što su žanr, popularnost izvođača,
sezonalnost i marketinški napori.

```python
df['ar_count'] = pd.Categorical(df['ar_count'])
```

```python
import seaborn as sns
import pandas as pd

sns.boxplot(x='ar_count', y='streams',hue='ar_count', data=df, palette='pastel')
plt.title("Distribucija broja izvođača pjesme i pregleda")
plt.xlabel("Broj izvođača")
plt.ylabel("Pregledi")
plt.show()
```

**Zaključak:** Grafikon prikazuje distribuciju pregleda pjesama u odnosu na broj izvođača koji učestvuju na
pjesmi. Ovo je prikazano pomoću boxplot dijagrama za svaki broj izvođača od 1 do 8, gdje je
'ar_count' legenda koja označava broj izvođača.
Iz boxplotova se može primijetiti sljedeće:
1. Pjesme s jednim izvođačem (plava kutija) imaju najveći raspon pregleda, sa srednjom
vrijednošću (medijanom) nižom od onih s više izvođača. Također, ima dosta izoliranih tačaka
iznad gornjeg kvartila, što označava da postoji znatan broj pjesama s jednim izvođačem koje
imaju izuzetno visok broj pregleda (outlieri).

2. Kako se broj izvođača povećava, medijan pregleda općenito opada. Tačnije, za pjesme sa 2
izvođača (narandžasta kutija) srednja vrijednost je viša nego za pjesme s jednim izvođačem, ali
maksimalni broj pregleda i raspon su manji.

3. Pjesme sa 3 i 4 izvođača (zelena i crvena kutija) pokazuju sličan trend s još nižim medijanima
i maksimalnim vrijednostima, iako ima manje izuzetno visokih vrijednosti u odnosu na pjesme
s manjim brojem izvođača.

4. Za pjesme s 5 do 8 izvođača (ljubičasta, žuta, svijetlo plava i siva kutija), medijani su
relativno niski i vrlo slični jedni drugima. Raspon pregleda i broj izuzetaka su također manji.
Ovo ukazuje na mogući zaključak da pjesme s manjim brojem izvođača, posebno solo izvođači,
imaju potencijal za veći broj pregleda. Međutim, prisustvo većeg broja izvođača na pjesmi ne
garantira veći broj pregleda i može čak ukazivati na nižu srednju vrijednost pregleda. Važno je
napomenuti da postojanje outliera može ukazivati na to da dok većina pjesama s jednim
izvođačem ima umjeren broj pregleda, postoji nekolicina takvih pjesama koje postižu iznimno
visoke brojke.

**HIPOTEZA 1-Broj pregleda (popularnost) pjesama varira ovisno o godišnjem dobu u kojem su objavljene.**

```python
import pandas as pd
import matplotlib.pyplot as plt

data_path = '/content/drive/MyDrive/spotify2023.csv'
df = pd.read_csv(data_path)
season_bins = [0, 3, 6, 9, 12]
season_labels = ['Winter', 'Spring', 'Summer', 'Autumn']

df['season'] = pd.cut(df['released_month'], bins=season_bins, labels=season_labels, include_lowest=True)

seasonal_popularity = df.groupby('season')['streams'].mean().reset_index()

plt.figure(figsize=(8, 6))
colors = ['blue', 'green', 'orange', 'grey']
plt.bar(seasonal_popularity['season'], seasonal_popularity['streams'], color=colors)
plt.xlabel('Godišnje doba')
plt.ylabel('Prosječna popularnost')
plt.title('Prosječna popularnost pjesama po godišnjem dobu')
plt.show()

```

**Zaključak:** Na osnovu grafikona, hipoteza da broj pregleda (popularnost) pjesama varira ovisno o godišnjem dobu u kojem su objavljene čini se potvrđenom. Grafikon pokazuje prosječnu popularnost pjesama za svako godišnje doba, i izgleda da postoje vidljive razlike u popularnosti. Pjesme objavljene u proljeće i jesen imaju manju popularnost u usporedbi s onima objavljenim u ljeto i zimu. Najpopularnije pjesme su one koje su objavljene u ljeto, dok su najmanje popularne one objavljene u proljeće. Ovi podaci sugeriraju da postoji sezonski trend u popularnosti pjesama, potkrepljujući navedenu hipotezu.

**Pitanje br.13:** *Postoji li statistički značajna veza između varijable "mode" (modus pjesme, npr. major ili minor) i varijable "energy_%" (energičnost pjesme)?*

*Za odgovor na ovo pitanje koristit ćemo Chi-kvadrat test kako bismo istražili postoji li povezanost između moda pjesme i njezine energičnosti.*

```python
import pandas as pd
from scipy.stats import chi2_contingency

df = pd.read_csv('/content/drive/MyDrive/spotify2023.csv', encoding='latin1')
contingency_table = pd.crosstab(df['mode'], pd.cut(df['energy_%'], bins=5))
chi2, p_value, dof, expected = chi2_contingency(contingency_table)

print("Rezultati Chi-kvadrat testa:")
print(f"Chi-kvadrat vrijednost: {chi2}")
print(f"P-vrijednost: {p_value}")
print(f"Stepeni slobode: {dof}")
print("Očekivane frekvencije:")
print(expected)

alpha = 0.05
if p_value < alpha:
    print("Postoji statistički značajna veza između varijable 'mode' i 'energy_%'.")
else:
    print("Nema statistički značajne veze između varijable 'mode' i 'energy_%'.")
```

**Zaključak:** Ovaj kod izračunava Chi-kvadrat test za varijable "mode" i "energy_%" kako bi istražio postoji li statistički značajna veza između moda pjesme i njezine energičnosti.

Rezultati Chi-kvadrat testa koje smo  dobili pokazuju da postoji statistički značajna veza između varijable 'mode' (modus pjesme) i varijable 'energy_%' (energičnost pjesme). P-vrijednost je manja od 0.05, što znači da imamo dovoljno statističkih dokaza da odbacimo nultu hipotezu i zaključimo da postoji značajna veza između ove dvije varijable.

Očekivane frekvencije prikazuju očekivanu raspodjelu frekvencija podataka za svaku kombinaciju moda i energetskih vrijednosti. Chi-kvadrat vrijednost mjeri koliko se stvarne frekvencije razlikuju od očekivanih frekvencija. Visoka Chi-kvadrat vrijednost ukazuje na veće odstupanje između stvarnih i očekivanih frekvencija, što doprinosi većoj statističkoj značajnosti.

Ovo je značajan rezultat koji sugerira da modus pjesme može utjecati na njezinu energetsku razinu, što može biti korisno za razumijevanje muzičke estetike i preferencija publike.

**Pitanje br.15:** *Proučavajući distribuciju broja streamova pjesama na Spotify platformi, koja je najčešća distribucija podataka - normalna, ravnomjerna ili možda asimetrična distribucija?*

*Kako bi ekstremne vrijednosti utjecale na prosječan broj streamova u odnosu na medijanu, te koji bi statistički metod bio najprikladniji za analizu ove distribucije?*

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from scipy import stats


df = pd.read_csv('/content/drive/MyDrive/spotify2023.csv', encoding='latin1')


plt.figure(figsize=(10, 6))
sns.histplot(df['streams'], bins=20, kde=True, color='skyblue')
plt.title('Distribucija broja streamova')
plt.xlabel('Broj streamova')
plt.ylabel('Frekvencija')
plt.show()

mean_streams = df['streams'].mean()
median_streams = df['streams'].median()

print("Srednja vrijednost broja streamova:", mean_streams)
print("Medijana broja streamova:", median_streams)
normality_test = stats.normaltest(df['streams'])
print("\nRezultati testa normalnosti:")
print("Statistika:", normality_test.statistic)
print("P-vrijednost:", normality_test.pvalue)
if normality_test.pvalue < 0.05:
    print("Distribucija nije normalna.")
else:
    print("Distribucija je normalna.")
if mean_streams > median_streams:
    print("\nSrednja vrijednost je veća od medijane.")
elif mean_streams < median_streams:
    print("\nMedijana je veća od srednje vrijednosti.")
else:
    print("\nSrednja vrijednost je jednaka medijani.")
```

Distribucija broja streamova pjesama na Spotify platformi nije normalna i ima asimetričnu formu iz nekoliko razloga:

  * Ekstremne vrijednosti: Postoje ekstremne visoke vrijednosti broja streamova koje povećavaju srednju vrijednost, čineći je većom od medijane. Ovo je karakteristika asimetrične distribucije.

  * Asimetrija distribucije: Vizualizacije distribucije broja streamova, poput histograma ili box plotova, pokazuju da oblik distribucije nije simetričan. Umjesto toga, distribucija se često proteže više prema desnoj strani, što ukazuje na asimetriju.

  * Statistički testovi normalnosti: Rezultati testa normalnosti pokazuju da p-vrijednost testa značajno odstupa od konvencionalne prage od 0.05. To znači da nema dovoljno dokaza da distribucija broja streamova slijedi normalnu distribuciju.

  * Realni scenariji: U stvarnom svijetu, distribucija broja streamova može biti pod utjecajem različitih čimbenika kao što su popularnost pojedinih pjesama, promocija, sezonske varijacije i tako dalje. Ovi faktori mogu doprinijeti asimetriji distribucije.

Sve ove karakteristike ukazuju na to da distribucija broja streamova pjesama na Spotify platformi nije normalna i da ima asimetričnu formu.

**Pitanje br.16:** *Pokazati kako Chebyshev- teoremom možemo dokazati koliki postotak pjesama leži unutar određenog raspona broja streamova od srednje vrijednosti.*

```python
import pandas as pd

df = pd.read_csv('/content/drive/MyDrive/spotify2023.csv', encoding='latin1')
mean_streams = df['streams'].mean()
std_dev_streams = df['streams'].std()

# Definiranje granica odstupanja
num_std_devs = 1
lower_bound = mean_streams - num_std_devs * std_dev_streams
upper_bound = mean_streams + num_std_devs * std_dev_streams

within_range = df[(df['streams'] >= lower_bound) & (df['streams'] <= upper_bound)].shape[0]
total_data_points = df.shape[0]
percentage_within_range = (within_range / total_data_points) * 100

print(f'Postotak pjesama unutar {num_std_devs} standardnih devijacija od srednje vrijednosti: {percentage_within_range:.2f}%')
```

**Pojašnjenje:** Chebyshev teorem nam pruža opću metodu za procjenu postotka podataka unutar određenog raspona od srednje vrijednosti, bez obzira na oblik distribucije podataka. Teorem kaže da je barem 1−1/k^2​ podataka unutar k standardnih devijacija od srednje vrijednosti.

Postotak pjesama unutar 1 standardne devijacije od srednje vrijednosti iznosi 85.31%. To znači da većina pjesama ima broj streamova koji se nalazi unutar jedne standardne devijacije od srednje vrijednosti. Ovo je važno sa aspekta razumijevanja distribucije popularnosti pjesama na Spotify platformi.
